```python
# Use a simple Perceptron to make predictions for multiple inputs.

import torch

# Multiple input samples.
x = torch.tensor([
    [1.0, 0.0],
    [0.0, 1.0],
    [1.0, 1.0],
    [0.0, 0.0]])

# One weight for each input feature.
weights = torch.tensor([0.5, -0.5])

# Bias shifts the decision boundary.
bias = torch.tensor(0.2)

# Calculate the weighted sum for every sample.
weighted_sum = torch.matmul(x, weights) + bias

# Convert weighted sums into binary predictions.
predictions = (weighted_sum >= 0).int()

print("Inputs:")
print(x)

print("\nWeighted Sums:")
print(weighted_sum)

print("\nPredictions:")
print(predictions)

# Explanation:
# The Perceptron applies the same weights and bias to every input sample.
# torch.matmul() calculates the weighted sum for each sample.
# The step condition then converts each result into either class 0 or 1.
```
