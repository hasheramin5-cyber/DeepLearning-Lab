# PyTorch Basics

PyTorch is an open-source machine learning framework widely used for building and training deep learning models.

In this notebook, we will explore the basic PyTorch workflow:

- Importing PyTorch
- Checking the installed PyTorch version
- Creating basic tensors
- Exploring tensor attributes
- Generating random tensors
- Managing CPU and GPU devices

The goal is to become familiar with the basic PyTorch environment before moving into detailed tensor operations.

```python
import torch

print("PyTorch imported successfully!")
print("PyTorch version:", torch.__version__)
```

## PyTorch Version

`torch.__version__` tells us which PyTorch version is currently installed in the environment.

```python
print("Installed PyTorch Version:", torch.__version__)
```

## Creating a Basic Tensor

A tensor is the fundamental data structure used by PyTorch.
It can represent a single value, a vector, a matrix, or higher-dimensional data.

```python
tensor = torch.tensor([10, 20, 30, 40, 50])

print("Tensor:")
print(tensor)
```

## Tensor Attributes

PyTorch tensors contain useful attributes that describe their data and structure.

Some important attributes and properties are:

- `dtype` --> data type of tensor values
- `shape` --> size of each dimension
- `ndim` --> number of dimensions
- `numel()` --> total number of elements
- `device` --> device where the tensor is stored

```python
print("Data Type:", tensor.dtype)
print("Shape:", tensor.shape)
print("Dimensions:", tensor.ndim)
print("Number of Elements:", tensor.numel())
print("Device:", tensor.device)
```

## Random Tensors

PyTorch can generate tensors containing random values.
`torch.rand()` generates random values from a uniform distribution between 0 and 1.

```python
random_tensor = torch.rand(3, 4)

print("Random Tensor:")
print(random_tensor)

print("\nShape:", random_tensor.shape)
print("Minimum Value:", random_tensor.min().item())
print("Maximum Value:", random_tensor.max().item())
```

## CPU and GPU Devices

PyTorch can perform computations on different devices.
If CUDA is available, tensors can be moved to a compatible NVIDIA GPU. Otherwise, PyTorch can use the CPU.
We can check CUDA availability using `torch.cuda.is_available()`.

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

tensor = torch.tensor([10.0, 20.0, 30.0]).to(device)

print("Selected Device:", device)
print("Tensor:", tensor)
print("Tensor Device:", tensor.device)

if torch.cuda.is_available():
    print("GPU Name:", torch.cuda.get_device_name(0))
else:
    print("CUDA is not available. Using CPU.")
```

## Basic PyTorch Workflow

The basic workflow explored in this notebook is:

1. Import PyTorch.
2. Check the installed version.
3. Create tensors.
4. Inspect tensor attributes.
5. Generate random tensors.
6. Select an available computation device.
7. Move tensors to the selected device.

These concepts provide the foundation for working with PyTorch before moving into detailed tensor operations and neural network development.

## Key Takeaways

- PyTorch provides tensors as its core data structure.
- `torch.__version__` shows the installed PyTorch version.
- Tensor properties such as `dtype`, `shape`, `ndim`, and `device` describe tensor data and structure.
- `torch.rand()` can generate random tensors with values between 0 and 1.
- `torch.cuda.is_available()` checks whether CUDA is available.
- `.to(device)` moves a tensor to the selected computation device.
