# Introduction to PyTorch

**PyTorch** is an open-source deep learning framework used to build, train, and deploy neural networks.
It provides powerful tools for working with tensors, automatic differentiation, neural network architectures, optimization, and GPU acceleration.

In this notebook, we will get familiar with PyTorch and perform a few basic operations before moving into tensors and model development.

## Importing PyTorch

The PyTorch library is imported using the `torch` module.
We will use `torch` throughout this notebook and the upcoming PyTorch experiments.

```python
import torch

print("PyTorch Version:", torch.__version__)
```

## Checking PyTorch Installation

We can check whether PyTorch can access a CUDA-enabled GPU.
If CUDA is unavailable, PyTorch will use the CPU instead.

```python
print("CUDA Available:", torch.cuda.is_available())

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
else:
    print("Using CPU")
```

## Creating a Basic Tensor

Tensors are the fundamental data structure used by PyTorch.
A tensor can represent scalars, vectors, matrices, and higher-dimensional data.

We will explore tensors in detail in the next notebook.

```python
x = torch.tensor([1, 2, 3, 4, 5])

print("Tensor:", x)
print("Shape:", x.shape)
print("Data Type:", x.dtype)
print("Number of Dimensions:", x.ndim)

# If you have explored the NumPy Practice folder, you may notice that many PyTorch operations are similar to NumPy.
# However, PyTorch provides additional features specifically designed for Deep Learning.
```

## Creating Random Data

PyTorch can generate tensors containing random values.
Random tensors are commonly used for initializing neural network weights and creating experimental data.

```python
random_tensor = torch.rand(3, 4)

print(random_tensor)
print("\nShape:", random_tensor.shape)
```

```python
# Have You Noticed That?

# torch.rand() generates random values between 0 and 1. The value 1 is not included in the generated range.
# If you want random values from a standard normal distribution, you can use torch.randn().

random_tensor = torch.randn(5, 7)

print(random_tensor)
print("\nShape:", random_tensor.shape)
```

## Tensor Computation

PyTorch tensors support mathematical operations directly.
These operations form the foundation for the numerical computations performed inside neural networks.

```python
a = torch.tensor([2.0, 4.0, 6.0])
b = torch.tensor([1.0, 3.0, 5.0])

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
```

## PyTorch and Deep Learning

PyTorch provides several important components for deep learning:

- Tensors for numerical data
- Autograd for automatic differentiation
- `torch.nn` for neural network layers and models
- Optimizers for updating model parameters
- Dataset and DataLoader utilities for handling data
- CUDA support for GPU acceleration
