# PyTorch Autograd

Autograd is PyTorch's automatic differentiation system.
It tracks operations performed on tensors and can automatically calculate gradients when needed.

In this notebook, we will explore:

- `requires_grad`
- Gradient calculation
- `.backward()`
- Gradient tracking
- `grad_fn`
- `torch.no_grad()`

```python
import torch

print("PyTorch Version:", torch.__version__)
```

## 1. Enabling Gradient Tracking

Setting `requires_grad=True` tells PyTorch to track operations performed on a tensor so that gradients can later be calculated.

```python
x = torch.tensor(5.0, requires_grad=True)

y = x ** 2

print("Input:", x)
print("Requires Gradient:", x.requires_grad)
print("Output:", y)
```

## 2. Calculating Gradients

The `.backward()` method calculates the gradient of the output with respect to tensors that have gradient tracking enabled.

For:

y = x²

the derivative is:

dy/dx = 2x

```python
x = torch.tensor(5.0, requires_grad=True)

y = x ** 2

y.backward()

print("Input:", x)
print("Output:", y)
print("Gradient:", x.grad)
```

## 3. The Backward Pass

The backward pass uses the computation performed on tensors to calculate gradients.

For:

y = x³

the derivative is:

dy/dx = 3x²

```python
x = torch.tensor(4.0, requires_grad=True)

y = x ** 3

y.backward()

print("Input:", x)
print("Output:", y)
print("Gradient:", x.grad)
```

## 4. Gradient Tracking

PyTorch keeps track of operations that involve tensors requiring gradients.
The `grad_fn` attribute can show the operation that created a tensor as part of the computation graph.

```python
x = torch.tensor(3.0, requires_grad=True)

y = x ** 2 + 2 * x + 1

print("Input:", x)
print("Output:", y)
print("Requires Gradient:", x.requires_grad)
print("Gradient Function:", y.grad_fn)

y.backward()

print("Gradient:", x.grad)
```

## 5. Disabling Gradient Tracking

`torch.no_grad()` temporarily disables gradient tracking.
This is useful when gradients are not required, such as during inference or when performing calculations that should not build a computation graph.

```python
x = torch.tensor(5.0, requires_grad=True)

y = x ** 2

print("Before disabling gradient tracking:")
print("Requires Gradient:", x.requires_grad)
print("Output:", y)

with torch.no_grad():
    z = x ** 2

print("\nInside no_grad() context:")
print("Output:", z)
print("Requires Gradient:", z.requires_grad)
```

## Key Takeaways

- `requires_grad=True` enables gradient tracking.
- `.backward()` calculates gradients.
- `.grad` stores the calculated gradient.
- `grad_fn` shows the operation that created a tracked tensor.
- PyTorch automatically builds the computation graph during tracked operations.
- `torch.no_grad()` disables gradient tracking temporarily.
