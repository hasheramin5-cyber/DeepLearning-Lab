```python
import torch

# Tell PyTorch to track operations on x
x = torch.tensor(2.0, requires_grad=True)

y = x ** 2     # Forward calculation
y.backward()   # Calculate the gradient

print("x:", x)
print("y:", y)
print("Gradient:", x.grad)


# Explanation:
# PyTorch calculates y = x² during the forward pass, backward() moves through the computational graph in reverse and calculates dy/dx. Since dy/dx = 2x and x = 2, the gradient is 4, which is stored in x.grad.
```

```python
import torch

# Track operations performed on x
x = torch.tensor(2.0, requires_grad=True)

# Build the computational graph
y = x**2 + 3*x + 5

# Calculate gradients through the graph
y.backward()

print("x:", x)
print("y:", y)
print("Gradient:", x.grad)


# Explanation:
# During the forward pass, PyTorch calculates y from x and tracks the operations used to get the result. backward() then moves through those operations in reverse and calculates the gradient of y with respect to x. For y = x² + 3x + 5, the gradient at x = 2 is 7.
```

```python
import torch

# Track gradients for both tensors
x = torch.tensor(2.0, requires_grad=True)
w = torch.tensor(3.0, requires_grad=True)

# Forward calculation
y = x * w

# Calculate gradients
y.backward()

print("x gradient:", x.grad)
print("w gradient:", w.grad)


# Explanation:
# PyTorch tracks the multiplication of x and w during the forward pass.
# backward() calculates how much each tensor affected y.
# Since dy/dx = w and dy/dw = x, the gradients are 3 and 2.
```

```python
import torch

# Track gradients for x
x = torch.tensor(2.0, requires_grad=True)

# Build a chain of operations
u = x ** 2
y = u * 3

# Calculate the gradient through the whole chain
y.backward()

print("x:", x)
print("y:", y)
print("Gradient:", x.grad)


# Explanation:
# PyTorch tracks both operations: x² and multiplication by 3.
# backward() applies the chain rule through these operations.
# The gradient is 12 because dy/dx = 3 × 2x and x = 2.
```

```python
import torch

# Track gradients for x
x = torch.tensor(2.0, requires_grad=True)

# First calculation and backward pass
y = x ** 2
y.backward()

print("After first backward:", x.grad)

# Clear the existing gradient
x.grad.zero_() # if u comment this line the pervious and next gradient would be added 

# Second calculation and backward pass
y = x ** 2
y.backward()

print("After second backward:", x.grad)


# Explanation:
# The first backward() calculates a gradient of 4.
# Gradients are stored in x.grad and can accumulate across backward passes.
# zero_() clears the existing gradient before calculating it again.
# This is why gradients are cleared during neural network training.
```

```python
# Without Clearing the Gradient

import torch

# Track gradients for x
x = torch.tensor(2.0, requires_grad=True)

# First calculation and backward pass
y = x ** 2
y.backward()

print("After first backward:", x.grad)

# Clear the existing gradient
x.grad.zero_()  # If this line is removed, the gradients from both backward passes accumulate.

# Second calculation and backward pass
y = x ** 2
y.backward()

print("After second backward:", x.grad)


# Explanation:
# The first backward() calculates a gradient of 4.
# Gradients are stored in x.grad and can accumulate across backward passes.
# zero_() clears the existing gradient before calculating it again.
```

```python
import torch

# Track gradients for x
x = torch.tensor(2.0, requires_grad=True)

# Normal operation is tracked
y = x ** 2

# Create a tensor detached from the graph
z = y.detach()

# Disable gradient tracking for this calculation
with torch.no_grad():
    a = x ** 2

print("y:", y)
print("z:", z)
print("a:", a)


# Explanation:
# y is connected to the computational graph because x requires gradients.
# detach() creates z without keeping its connection to the graph.
# no_grad() temporarily disables gradient tracking for calculations inside it.
# Both are useful when gradients are not needed, especially during inference.
```


**`detach()`**
   →
Separates a specific tensor from the computational graph.

**`no_grad()`**
   →
Temporarily disables gradient tracking for all operations inside the block.


```python
import torch

# Track operations on x
x = torch.tensor(2.0, requires_grad=True)

# Build a chain of operations
y = x ** 2
z = y * 3

z.backward()

print(x.grad)

print("y:", y)
print("y grad_fn:", y.grad_fn)

print("z:", z)
print("z grad_fn:", z.grad_fn)

# Explanation:
# Because x requires gradients, PyTorch tracks the operations used to calculate y and z.
# grad_fn shows the backward operation PyTorch has stored for calculating gradients later.
```

```python
import torch

# Input and weight need gradients
x = torch.tensor(2.0, requires_grad=True)
w = torch.tensor(3.0, requires_grad=True)

# Forward pass
prediction = x * w

# Calculate the loss
target = torch.tensor(10.0)
loss = (prediction - target) ** 2

# Backpropagation
loss.backward()

print("Prediction:", prediction)
print("Loss:", loss)
print("Weight Gradient:", w.grad)


# Explanation:
# The forward pass calculates the prediction and then the loss.
# backward() moves through the computational graph in reverse and calculates how the weight affected the loss.
# The resulting gradient is stored in w.grad and will later be used by an optimizer to update the weight.
```

```python
import torch

# Track operations on x
x = torch.tensor(2.0, requires_grad=True)

# Forward calculation
y = x ** 2

# Calculate the gradient directly
gradient = torch.autograd.grad(y, x)

print("Gradient:", gradient)
print("Gradient:", gradient[0])


# Explanation:
# autograd.grad() calculates the derivative of y with respect to x and returns the gradient directly instead of storing it in x.grad.
# For y = x² at x = 2, the gradient is 4.
```
