# Optimizers

Optimizers are algorithms that update the weights and biases of a Neural Network to minimize the loss function.
Their objective is to find the optimal values of the model parameters that produce the lowest possible error during training.

## Why Do We Need Optimizers?

After the Forward Pass, the model calculates a prediction.
The Loss Function measures how far the prediction is from the actual value.
The Optimizer uses this loss to update the weights and biases so the model performs better in the next iteration.

Without an optimizer, a Neural Network cannot learn.

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Stochastic Gradient Descent (SGD)

Stochastic Gradient Descent (SGD) updates the model parameters using the gradient of the loss function.
It is one of the simplest and most commonly used optimization algorithms.

```python
epochs = np.arange(1, 51)

sgd_loss = 1.2 * np.exp(-epochs / 8) + 0.05

plt.figure(figsize=(8,5))

plt.plot(
    epochs,
    sgd_loss,
    marker = "o",
    markerfacecolor = "blue",
    markeredgecolor = "blue",
    markersize = 4,
    color="red",
    linewidth=2,
    label="SGD"
)

plt.title("SGD Optimizer")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.grid(True)
plt.legend()

plt.savefig(
    "Assets/Optimizers/sgd_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

## Adam Optimizer

Adam (Adaptive Moment Estimation) combines momentum and adaptive learning rates.
It generally converges faster and is one of the most popular optimizers used in Deep Learning.

```python
adam_loss = 1.2 * np.exp(-epochs / 15) + 0.02

plt.figure(figsize=(8,5))

plt.plot(
    epochs,
    adam_loss,
    marker = "o",
    markersize = 4,
    markerfacecolor = "blue",
    markeredgecolor = "blue",
    color="red",
    linewidth=2,
    label="Adam"
)

plt.title("Adam Optimizer")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.grid(True)
plt.legend()

plt.savefig(
    "Assets/Optimizers/adam_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

## SGD vs Adam

The graph below compares the convergence of SGD and Adam.
Adam usually reaches a lower loss in fewer epochs, while SGD may require more iterations to converge.

```python
plt.figure(figsize=(8,5))

plt.plot(
    epochs,
    sgd_loss,
    marker = "o",
    markerfacecolor = "lightgray",
    markeredgecolor = "#000000",
    markersize = 4,
    linewidth=1,
    color = "blue",
    label="SGD"
)

plt.plot(
    epochs,
    adam_loss,
    marker = "o",
    markerfacecolor = "lightgray",
    markeredgecolor = "#000000",
    markersize = 4,
    linewidth=1,
    color = "#26ff00",
    label="Adam"
)

plt.title("SGD vs Adam")
plt.xlabel("Epoch")
plt.ylabel("Loss")

plt.grid(True)

plt.legend()

plt.savefig(
    "Assets/Optimizers/sgd_vs_adam_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

## Key Takeaways

- Optimizers update the weights and biases of a Neural Network.
- SGD is simple and computationally efficient.
- Adam combines momentum with adaptive learning rates.
- Adam generally converges faster than SGD.
- Choosing the right optimizer can significantly improve training performance.
