```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Sigmoid Activation Function

x = np.linspace(-10, 10, 500)
sigmoid = 1 / (1 + np.exp(-x))

plt.figure(figsize=(8,5))
plt.plot(x, sigmoid, linewidth=2, label="Sigmoid")
plt.title("Sigmoid Activation Function")
plt.xlabel("Input")
plt.ylabel("Output")
plt.grid(True)
plt.legend()
plt.savefig("Assets/Activation Functions/sigmoid_graph.png",
            dpi=300,
            bbox_inches="tight")
plt.show()
```

```python
# ReLU Activation Function

x = np.linspace(-10, 10, 500)
relu = np.maximum(0, x)

plt.figure(figsize=(8,5))
plt.plot(x, relu, linewidth=2, label="ReLU")
plt.title("ReLU Activation Function")
plt.xlabel("Input")
plt.ylabel("Output")
plt.grid(True)
plt.legend()
plt.savefig("Assets/Activation Functions/relu_graph.png",
            dpi=300,
            bbox_inches="tight")
plt.show()
```

```python
# Leaky ReLU Activation Function

x = np.linspace(-10, 10, 500)
alpha = 0.01
leaky_relu = np.where(x > 0, x, alpha * x)

plt.figure(figsize=(8,5))
plt.plot(x, leaky_relu, linewidth=2, label="Leaky ReLU")
plt.title("Leaky ReLU Activation Function")
plt.xlabel("Input")
plt.ylabel("Output")
plt.grid(True)
plt.legend()
plt.savefig("Assets/Activation Functions/leaky_relu_graph.png",
            dpi=300,
            bbox_inches="tight")
plt.show()
```

```python
# Softmax Activation Function

x = np.linspace(-5, 5, 200)
softmax = np.exp(x) / np.sum(np.exp(x))

plt.figure(figsize=(8,5))
plt.plot(x, softmax, linewidth=2, label="Softmax")
plt.title("Softmax Activation Function")
plt.xlabel("Input")
plt.ylabel("Probability")
plt.grid(True)
plt.legend()
plt.savefig("Assets/Activation Functions/softmax_graph.png",
            dpi=300,
            bbox_inches="tight")
plt.show()
```

```python
# Softmax Activation Function Bar Chart

scores = np.array([2.0, 1.0, 0.1])

exp_scores = np.exp(scores - np.max(scores))
softmax = exp_scores / np.sum(exp_scores)

print("Scores       :", scores)
print("Probabilities:", softmax)

print()
plt.figure(figsize=(6,4))
plt.bar(["Class 1", "Class 2", "Class 3"], softmax)
plt.title("Softmax Output Probabilities")
plt.ylabel("Probability")
plt.ylim(0, 1)
plt.grid(axis="y")
plt.savefig("Assets/Activation Functions/softmax_bar-graph.png",
            dpi=300,
            bbox_inches="tight")
plt.show()
```
