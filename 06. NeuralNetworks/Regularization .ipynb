# Regularization

Regularization is a technique used to reduce overfitting in Machine Learning and Deep Learning models.
It helps the Neural Network generalize better by preventing it from memorizing the training data.

## Why Regularization?

A Neural Network may perform very well on training data but poorly on unseen data.
This problem is called **Overfitting**.
Regularization reduces model complexity and improves generalization.

```python
import numpy as np
import matplotlib.pyplot as plt
```

## L1 - Regularization

L1 Regularization adds the absolute value of the weights to the loss function.
It encourages some weights to become zero, making the model simpler.
It is also known as **Lasso Regularization**.

```python
weights = np.array([0.8, -0.4, 1.2, -0.6])
l1_penalty = np.sum(np.abs(weights))

print("Weights :", weights)
print("L1 Penalty :", l1_penalty)
```

## L2 - Regularization

L2 Regularization adds the squared value of the weights to the loss function.
Instead of removing weights completely, it reduces their magnitude.
It is also known as **Ridge Regularization**.

```python
l2_penalty = np.sum(weights**2)

print("Weights :", weights)
print("L2 Penalty :", round(l2_penalty,2))
```

## Dropout

Dropout randomly disables a portion of neurons during training.
This prevents neurons from depending too much on each other and improves generalization.
Dropout is only active during training and is disabled during testing.

```python
epochs = np.arange(1,51)

without_dropout = 1.2*np.exp(-epochs/14)+0.02
with_dropout = 1.3*np.exp(-epochs/11)+0.08

plt.figure(figsize=(8,5))

plt.plot(
    epochs,
    without_dropout,
    linewidth=2.5,
    color = "red",
    label="Without Dropout")

plt.plot(
    epochs,
    with_dropout,
    linewidth=2.5,
    color = "blue",
    label="With Dropout")

plt.title(
    "Regularization Effect")

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.grid(axis = "both",
         linestyle = "dashed",
         color = "lightgray")
plt.legend()
plt.savefig(
    "Assets/Regularization/regularization_graph.png",
    dpi=300,
    bbox_inches="tight")

plt.show()
```

## Key Takeaways

- Regularization reduces overfitting.
- L1 Regularization can make some weights exactly zero.
- L2 Regularization keeps weights small and stable.
- Dropout randomly disables neurons during training.
- Regularization improves the model's ability to generalize to unseen data.
