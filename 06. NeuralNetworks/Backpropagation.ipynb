# Backpropagation

Backpropagation is the learning process of a Neural Network.
After making a prediction, the network calculates the error and sends it backward through the network to update the weights and biases.

This process helps the model improve its predictions over time.

## Why Backpropagation?

A Forward Pass only generates a prediction.
If the prediction is incorrect, the model must learn from its mistake.
Backpropagation calculates how much each weight contributed to the error and updates those weights to reduce future errors.

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Step 1 - Prediction

Suppose our Neural Network predicts a value.
We compare this prediction with the actual value.

```python
actual = 8
prediction = 6.5

print("Actual Value :", actual)
print("Predicted Value :", prediction)
```

## Step 2 - Calculate Loss

The loss measures how far the prediction is from the actual value.
Here we use Mean Squared Error (MSE).

```python
loss = (actual - prediction) ** 2
print("Loss :", loss)
```

## Step 3 - Weight Update

During Backpropagation, the optimizer adjusts the weights.
The updated weights help reduce the loss in the next Forward Pass.

```python
old_weight = 0.50
gradient = -0.15
learning_rate = 0.10

new_weight = old_weight - learning_rate * gradient

print("Old Weight :", old_weight)
print("Updated Weight :", round(new_weight,3))
```

## Backpropagation Visualization

The graph below illustrates how the training loss decreases after multiple weight updates.

```python
epochs = np.arange(1,51)

loss = 1.1*np.exp(-epochs/10)+0.03

plt.figure(figsize=(8,5))

plt.plot(
    epochs,
    loss,
    linewidth=2.5,
    color="red")

plt.scatter(
    epochs[::5],
    loss[::5],
    color="black",
    s=30)

plt.title(
    "Backpropagation Learning",
    fontsize=16,
    fontweight="bold")

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.grid(
    linestyle="dashed",
    alpha=0.5)

plt.savefig(
    "Assets/Backpropagation/backpropagation_graph.png",
    dpi=300,
    bbox_inches="tight")

plt.show()
```

## Key Takeaways

- Backpropagation helps a Neural Network learn from its mistakes.
- The prediction error is calculated using a Loss Function.
- Gradients measure how much each weight contributed to the error.
- The optimizer updates the weights using these gradients.
- Forward Pass and Backpropagation work together during every training iteration.
