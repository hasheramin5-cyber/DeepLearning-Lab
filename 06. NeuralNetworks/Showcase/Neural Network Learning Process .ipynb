# Neural Network Learning Process

This notebook visualizes three fundamental concepts of Deep Learning:

- Activation Function (ReLU)
- Training Loss
- Optimizer Comparison (SGD vs Adam)

Built using NumPy and Matplotlib.

```python
import numpy as np
import matplotlib.pyplot as plt

# Data
x = np.linspace(-5, 5, 400)
relu = np.maximum(0, x)

epochs = np.arange(1, 51)

loss = 1.2 * np.exp(-epochs / 12) + 0.03
sgd = 1.2 * np.exp(-epochs / 8) + 0.05
adam = 1.2 * np.exp(-epochs / 15) + 0.02

# Figure
fig, axes = plt.subplots(1,
                         3,
                         figsize=(22,7),
                         facecolor="#f8f9fa")

fig.suptitle("Neural Network Learning Process",
             fontsize=24,
             fontweight="bold")

# ReLU
axes[0].plot(x,
             relu,
             color="#0066CC",
             linewidth=3)

axes[0].set_title("Activation Function (ReLU)",
                  fontsize=16,
                  fontweight="bold")

axes[0].set_xlabel("Input")
axes[0].set_ylabel("Output")
axes[0].grid(linestyle="--", alpha=0.5)

# Loss
axes[1].plot(epochs,
             loss,
             color="#28A745",
             linewidth=3,
             marker="o",
             markersize=4,
             markevery=4)

axes[1].set_title("Training Loss",
                  fontsize=16,
                  fontweight="bold")

axes[1].set_xlabel("Epoch")
axes[1].set_ylabel("Loss")
axes[1].grid(linestyle="--", alpha=0.5)

# Optimizers
axes[2].plot(epochs,
             sgd,
             color="#E74C3C",
             linewidth=3,
             marker="o",
             markersize=4,
             markevery=4,
             label="SGD")

axes[2].plot(epochs,
             adam,
             color="#8E44AD",
             linewidth=3,
             marker="s",
             markersize=4,
             markevery=4,
             label="Adam")

axes[2].set_title("Optimizer Comparison",
                  fontsize=16,
                  fontweight="bold")

axes[2].set_xlabel("Epoch")
axes[2].set_ylabel("Loss")
axes[2].grid(linestyle="--", alpha=0.5)
axes[2].legend(frameon=True, fontsize=11)

# Bottom Caption
fig.text(0.5, 0.02,
         "Activation → Prediction → Loss → Optimizer → Better Model",
         ha="center",
         fontsize=13,
         fontweight="bold",
         color="gray")

plt.subplots_adjust(left=0.05,
                    right=0.98,
                    top=0.84,
                    bottom=0.12,
                    wspace=0.32)

plt.savefig("Assets/neural_network_learning_process.png",
            dpi=300,
            bbox_inches="tight")

plt.show()

# This visualization is created for educational and portfolio purposes.
```
