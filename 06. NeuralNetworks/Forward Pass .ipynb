# Forward Pass

A Forward Pass is the process where input data travels through the Neural Network to produce a prediction.
Each neuron performs a weighted sum, adds a bias, applies an activation function, and passes the result to the next layer.

## Forward Pass Workflow

A Forward Pass follows these steps:

1. Receive Inputs
2. Multiply Inputs by Weights
3. Add Bias
4. Apply Activation Function
5. Produce Output

Every prediction made by a Neural Network starts with a Forward Pass.

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Step 1 - Inputs

The network receives numerical input values.
These values represent the information that will be processed by the neuron.

```python
inputs = np.array([2,4,6])
print("Inputs :", inputs)
```

## Step 2 - Weighted Sum

Each input is multiplied by its corresponding weight.
The weighted values are added together along with the bias.

```python
weights = np.array([0.5,0.2,0.8])

bias = 1
weighted_sum = np.dot(inputs,weights)+bias
print("Weighted Sum :",weighted_sum)
```

## Step 3 - Activation Function

The weighted sum is passed through an activation function.
Here we use the ReLU activation function.

```python
output = np.maximum(0,weighted_sum)
print("Neuron Output :",output)
```

## Forward Pass Visualization

The graph below shows how information moves through a single neuron during a Forward Pass.

```python
steps = [
    "Input",
    "Weighted\nSum",
    "Bias",
    "ReLU",
    "Output"]

values = [
    3,
    weighted_sum,
    bias,
    output,
    output]

plt.figure(figsize=(8,5))

bars = plt.bar(
    steps,
    values,
    color=[
        "#3498DB",
        "#2ECC71",
        "#F39C12",
        "#9B59B6",
        "#E74C3C"])

plt.title(
    "Forward Pass Flow",
    fontsize=16,
    fontweight="bold")

plt.ylabel("Value")
plt.grid(
    axis="y",
    linestyle="dashed",
    alpha=0.5)

for bar in bars:
    plt.text(
        bar.get_x()+bar.get_width()/2,
        bar.get_height()+0.15,
        f"{bar.get_height():.1f}",
        ha="center",
        fontweight="bold")

plt.savefig(
    "Assets/Forward Pass/forward_pass_graph.png",
    dpi=300,
    bbox_inches="tight")

plt.show()
```

## Key Takeaways

- A Forward Pass generates predictions.
- Inputs are multiplied by weights.
- Bias is added to the weighted sum.
- The activation function produces the final output.
- Every Neural Network prediction begins with a Forward Pass.
