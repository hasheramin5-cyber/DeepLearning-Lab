# Artificial Neuron

An Artificial Neuron is the fundamental building block of a Neural Network.

It receives multiple input values, multiplies each input by its corresponding weight, adds a bias, and then passes the result through an activation function to produce the final output.

This notebook demonstrates how a single artificial neuron works using NumPy.

## Mathematical Representation

A single artificial neuron computes:

Output = Activation((x₁·w₁) + (x₂·w₂) + ... + (xₙ·wₙ) + b)

where:

- x = Inputs
- w = Weights
- b = Bias

```python
import numpy as np
```

```python
# Inputs
inputs = np.array([2, 4, 6])

# Weights
weights = np.array([0.5, 0.2, 0.8])

# Bias
bias = 1

weighted_sum = np.dot(inputs, weights) + bias
print("Weighted Sum :", weighted_sum)
```

```python
# ReLU Activation

output = np.maximum(0, weighted_sum)
print("Neuron Output :", output)
```

```python
inputs = np.array([1, 3, 5])

weighted_sum = np.dot(inputs, weights) + bias
output = np.maximum(0, weighted_sum)
print(output)
```

```python
weights = np.array([1.0, 0.3, 0.1])

weighted_sum = np.dot(inputs, weights) + bias
output = np.maximum(0, weighted_sum)
print(output)
```

```python
bias = -5

weighted_sum = np.dot(inputs, weights) + bias
output = np.maximum(0, weighted_sum)
print(output)
```

## Key Takeaways

- Inputs carry information.
- Weights determine the importance of each input.
- Bias shifts the weighted sum.
- Activation Function decides the final output.
- Millions of these neurons together form a Deep Neural Network.
