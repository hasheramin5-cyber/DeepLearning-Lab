# Loss Functions Visualization

In this notebook, we will visualize the most commonly used Loss Functions in Deep Learning.

Topics Covered:

- Mean Squared Error (MSE)
- Binary Cross Entropy (BCE)
- Categorical Cross Entropy (CCE)
- Prediction Confidence vs Loss
- Comparison of MSE and BCE

Libraries Used:

- NumPy
- Matplotlib

```python
# Mean Squared Error (MSE) Visualization

import numpy as np
import matplotlib.pyplot as plt


actual = 1

predictions = np.linspace(0, 1, 200)
mse_loss = (actual - predictions) ** 2

plt.figure(figsize=(8,5))
plt.plot(predictions, mse_loss, linewidth = 3, color = "blue", linestyle = "dotted")

plt.title("Mean Squared Error (MSE)")
plt.xlabel("Prediction")
plt.ylabel("Loss")

plt.grid(True)

plt.show()
```

```python
# Binary Cross Entropy (BCE) Visualization

import numpy as np
import matplotlib.pyplot as plt


actual = 1

predictions = np.linspace(0.001, 0.999, 300)
bce_loss = -np.log(predictions)

plt.figure(figsize=(8,5))
plt.plot(predictions, bce_loss, linewidth = 3, color = "blue", linestyle = "dotted")

plt.title("Binary Cross Entropy Loss")
plt.xlabel("Predicted Probability")
plt.ylabel("Loss")

plt.grid(True)

plt.show()
```

```python
# Categorical Cross Entropy Example

import numpy as np
import matplotlib.pyplot as plt


probabilities = np.array([0.90,0.70,0.50,0.30,0.10])

loss = -np.log(probabilities)

plt.figure(figsize=(8,5))

plt.plot(probabilities, loss, marker = "o", markerfacecolor = "red", markeredgecolor = "red", linewidth = 3, color = "blue")

plt.title("Categorical Cross Entropy")
plt.xlabel("Correct Class Probability")
plt.ylabel("Loss")

plt.grid(True)

plt.show()
```

```python
# Correct Prediction vs Wrong Prediction

import matplotlib.pyplot as plt

labels = ["Correct\nPrediction","Wrong\nPrediction"]
losses = [0.05,2.30]

plt.figure(figsize=(6,5))

plt.bar(labels,losses)

plt.title("Correct vs Wrong Prediction")
plt.ylabel("Loss")

plt.grid(axis="y")

plt.show()
```

```python
# Prediction Confidence vs Loss

import numpy as np
import matplotlib.pyplot as plt


confidence = np.linspace(0.01,0.99,300)

loss = -np.log(confidence)

plt.figure(figsize=(8,5))

plt.plot(confidence, loss, linewidth = 3, color = "red")

plt.title("Prediction Confidence vs Loss")
plt.xlabel("Prediction Confidence")
plt.ylabel("Loss")

plt.grid(True)

plt.show()
```

```python
# Binary Cross Entropy Curve

import numpy as np
import matplotlib.pyplot as plt


p = np.linspace(0.001,0.999,500)

loss_positive = -np.log(p)
loss_negative = -np.log(1-p)

plt.figure(figsize=(9,5))

plt.plot(p, loss_positive, label = "Actual = 1", linewidth = 3, color = "blue")
plt.plot(p, loss_negative, label = "Actual = 0", linewidth = 3, color = "red")

plt.title("Binary Cross Entropy Curves")
plt.xlabel("Predicted Probability")
plt.ylabel("Loss")

plt.grid(True)
plt.legend()

plt.show()
```

```python
# MSE vs BCE Comparison

import numpy as np
import matplotlib.pyplot as plt


prediction = np.linspace(0.001,0.999,300)

mse = (1-prediction)**2
bce = -np.log(prediction)

plt.figure(figsize=(9,5))

plt.plot(prediction,mse,label="MSE",linewidth=3, color = "red")
plt.plot(prediction,bce,label="Binary Cross Entropy",linewidth=3, color = "blue")

plt.title("MSE vs Binary Cross Entropy")
plt.xlabel("Prediction")
plt.ylabel("Loss")

plt.grid(axis = "both", color = "lightgray", linestyle = "dashed")
plt.legend()

plt.show()
```

# Notebook Explanation

The purpose of this notebook is to build an intuitive understanding of how different loss functions measure prediction errors during model training.

---

## Program 1 - Mean Squared Error (MSE)

- MSE calculates the squared difference between the actual value and the predicted value.
- As the prediction gets closer to the actual value, the loss approaches zero.
- MSE is mainly used for **Regression Problems**.

---

## Program 2 - Binary Cross Entropy (BCE)

- BCE is commonly used in **Binary Classification**.
- As the predicted probability for the correct class approaches **1**, the loss rapidly decreases.
- Wrong predictions receive a much larger penalty than MSE.

---

## Program 3 - Categorical Cross Entropy (CCE)

- CCE is used with **Softmax** for Multi-Class Classification.
- Higher probability assigned to the correct class results in lower loss.
- Lower confidence results in significantly higher loss.

---

## Program 4 - Correct Prediction vs Wrong Prediction

This visualization demonstrates:

- Correct predictions produce very small loss.
- Wrong predictions produce much larger loss.
- During training, the optimizer tries to minimize this loss.

---

## Program 5 - Prediction Confidence vs Loss

This graph clearly shows:

- Higher confidence → Lower loss.
- Lower confidence → Higher loss.

The relationship is not linear; the loss changes rapidly for poor predictions.

---

## Program 6 - Binary Cross Entropy Curves

This graph displays BCE behavior for both possible classes.

- If the actual label is **1**, predicting values close to **1** minimizes loss.
- If the actual label is **0**, predicting values close to **0** minimizes loss.

This visualization explains why BCE works effectively for binary classification.

---

## Program 7 - MSE vs Binary Cross Entropy

This comparison highlights the difference between the two loss functions.

- MSE changes more smoothly.
- BCE penalizes incorrect classification much more aggressively.
- This is one of the main reasons why Cross Entropy is preferred for classification tasks.

---

## Final Summary

| Loss Function | Used With | Used For |
|--------------|-----------|-----------|
| MSE | Linear Output | Regression |
| BCE | Sigmoid | Binary Classification |
| CCE | Softmax | Multi-Class Classification |

---

# Key Takeaways

- **MSE** → Regression Problems
- **Binary Cross Entropy** → Binary Classification
- **Categorical Cross Entropy** → Multi-Class Classification
- Better predictions always produce lower loss.
- Cross Entropy provides stronger learning signals for classification tasks than MSE.
