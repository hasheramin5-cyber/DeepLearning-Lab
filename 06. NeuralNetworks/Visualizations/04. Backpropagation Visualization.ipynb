```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Initial Values
x = 0.8                 # Input Feature
y = 1                   # Actual Label (Cat)

weight = 0.50
bias = 0.20

learning_rate = 0.10
epochs = 10

history = []

# Training Loop
for epoch in range(1, epochs + 1):

    # Forward Pass
    z = weight * x + bias
    prediction = sigmoid(z)

    # Binary Cross Entropy Loss
    loss = -(y * np.log(prediction) + (1 - y) * np.log(1 - prediction))

    # Backpropagation
    gradient = prediction - y

    dW = gradient * x
    dB = gradient

    # Optimizer Step
    weight -= learning_rate * dW
    bias -= learning_rate * dB

    history.append([
        epoch,
        round(prediction,4),
        round(loss,4),
        round(weight,4),
        round(bias,4)
    ])

# Graph
epochs_list = [row[0] for row in history]
loss_list = [row[2] for row in history]

plt.figure(figsize=(8,5))

plt.plot(
    epochs_list,
    loss_list,
    marker="o",
    markerfacecolor = "#03fc13",
    markeredgecolor = "blue",
    linewidth=2,
    color = "blue"
)

plt.title("Backpropagation Learning")
plt.xlabel("Epoch")
plt.ylabel("Loss")

plt.grid(True)
plt.show()

# Training Summary Table
table = pd.DataFrame(
    history,
    columns=[
        "Epoch",
        "Prediction",
        "Loss",
        "Weight",
        "Bias"
    ]
)

print("\nTraining Summary:\n")

print(table)
```

# Notebook Explanation

This notebook demonstrates how a Neural Network improves during training using **Backpropagation**.

The graph shows how the **Loss decreases** as the number of epochs increases. A decreasing loss indicates that the model is learning from its mistakes and making better predictions.

The table provides a summary of each training epoch:

- **Epoch** → Current training iteration.
- **Prediction** → Output probability produced during the Forward Pass.
- **Loss** → Error measured using Binary Cross Entropy.
- **Weight** → Updated model parameter after Backpropagation.
- **Bias** → Updated bias after each optimization step.

Notice that after every epoch:

- The prediction becomes more accurate.
- The loss gradually decreases.
- The weights and bias are updated using the calculated gradients.

This process continues until the model converges and produces accurate predictions.
