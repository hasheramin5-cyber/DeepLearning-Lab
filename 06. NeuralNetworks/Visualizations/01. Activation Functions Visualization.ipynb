# Activation Functions Visualization

In this notebook, we will visualize the most commonly used activation functions in Deep Learning.

Topics Covered:

- Sigmoid
- ReLU
- Leaky ReLU
- Softmax
- Comparison of Activation Functions

Libraries Used:

- NumPy
- Matplotlib

```python
# Sigmoid Visualization

import numpy as np
import matplotlib.pyplot as plt


def sigmoid(x):
    return 1 / (1 + np.exp(-x))


x = np.linspace(-10, 10, 500)
y = sigmoid(x)

plt.figure(figsize=(8,5))
plt.plot(x, y, linewidth=3, label="Sigmoid", color = "blue")

plt.title("Sigmoid Activation Function")
plt.xlabel("Input (x)")
plt.ylabel("Output")

plt.grid(True)
plt.legend()

plt.show()
```

```python
# ReLU Visualization

import numpy as np
import matplotlib.pyplot as plt


def relu(x):
    return np.maximum(0, x)


x = np.linspace(-10,10,500)
y = relu(x)

plt.figure(figsize=(8,5))
plt.plot(x, y, linewidth = 3, label = "ReLU", color = "blue")

plt.title("ReLU Activation Function")
plt.xlabel("Input (x)")
plt.ylabel("Output")

plt.grid(True)
plt.legend()

plt.show()
```

```python
# Leaky ReLU Visualization

import numpy as np
import matplotlib.pyplot as plt


def leaky_relu(x):
    return np.where(x > 0, x, 0.01 * x)


x = np.linspace(-10,10,500)
y = leaky_relu(x)

plt.figure(figsize=(8,5))
plt.plot(x, y, linewidth = 3, label = "Leaky ReLU", color = "blue")

plt.title("Leaky ReLU Activation Function")
plt.xlabel("Input (x)")
plt.ylabel("Output")

plt.grid(True)
plt.legend()

plt.show()
```

```python
# Sigmoid vs ReLU

import numpy as np
import matplotlib.pyplot as plt


def sigmoid(x):
    return 1/(1+np.exp(-x))

def relu(x):
    return np.maximum(0,x)


x=np.linspace(-10,10,500)

plt.figure(figsize=(9,5))

plt.plot(x, sigmoid(x), linewidth = 3, label = "Sigmoid", color = "blue")
plt.plot(x, relu(x), linewidth = 3, label = "ReLU", color = "red")

plt.title("Sigmoid vs ReLU")
plt.xlabel("Input")
plt.ylabel("Output")

plt.grid(True)
plt.legend()

plt.show()
```

```python
# ReLU vs Leaky ReLU

import numpy as np
import matplotlib.pyplot as plt


def relu(x):
    return np.maximum(0,x)

def leaky_relu(x):
    return np.where(x>0,x,0.01*x)


x=np.linspace(-10,10,500)

plt.figure(figsize=(9,5))

plt.plot(x, relu(x), linewidth = 3, label = "ReLU", color = "blue")
plt.plot(x, leaky_relu(x), linewidth = 3, label = "Leaky ReLU", color = "red")

plt.title("ReLU vs Leaky ReLU")
plt.xlabel("Input")
plt.ylabel("Output")

plt.grid(True)
plt.legend()

plt.show()
```

```python
# Softmax Visualization

import numpy as np
import matplotlib.pyplot as plt


scores = np.array([2.5, 1.2, 0.3, 4.8])

exp_scores = np.exp(scores - np.max(scores))
probabilities = exp_scores / np.sum(exp_scores)

classes = ["Cat","Dog","Horse","Bird"]

plt.figure(figsize=(8,5))

plt.bar(classes, probabilities, color = "blue")

plt.title("Softmax Probabilities")
plt.xlabel("Classes")
plt.ylabel("Probability")

plt.grid(axis="y")

plt.show()

print("Softmax Output:")

for c,p in zip(classes,probabilities):
    print(f"{c} : {p:.4f}")
```

```python
# All Activation Functions Together

import numpy as np
import matplotlib.pyplot as plt


def sigmoid(x):
    return 1/(1+np.exp(-x))

def relu(x):
    return np.maximum(0,x)

def leaky_relu(x):
    return np.where(x>0,x,0.01*x)


x=np.linspace(-10,10,500)

plt.figure(figsize=(10,6))

plt.plot(x, sigmoid(x), linewidth = 3, label = "Sigmoid", color = "blue")
plt.plot(x, relu(x), linewidth = 3, label = "ReLU", color = "red")
plt.plot(x, leaky_relu(x), linewidth = 3, label = "Leaky ReLU", color = "#00ff1a")

plt.title("Activation Functions Comparison")
plt.xlabel("Input")
plt.ylabel("Output")

plt.grid(True)
plt.legend()

plt.show()
```

# Notebook Explanation

The purpose of this notebook was not only to plot activation function graphs but also to build an intuitive understanding of how each activation function behaves and why it is used in Deep Learning.

---

## Program 1 - Sigmoid Activation Function

- The output is always compressed between **0 and 1**.
- This makes Sigmoid suitable for **Binary Classification** problems.
- Extremely large positive and negative values are compressed near **1** and **0**, respectively.

---

## Program 2 - ReLU Activation Function

- All **negative values become 0**.
- All **positive values remain unchanged**.
- Because ReLU does not compress positive values, it helps Deep Neural Networks learn much faster.

---

## Program 3 - Leaky ReLU Activation Function

- Similar to ReLU, but it does **not completely discard negative values**.
- A small slope is kept for negative inputs.
- This helps prevent the **Dying ReLU Problem**, allowing neurons to continue learning.

---

## Program 4 - Sigmoid vs ReLU

This comparison clearly shows that:

- **Sigmoid** compresses every value into the range **0 to 1**.
- **ReLU** keeps positive values unchanged while removing negative values.

This is one of the most common interview comparison graphs.

---

## Program 5 - ReLU vs Leaky ReLU

This visualization highlights the main difference between the two functions.

- **ReLU** completely blocks negative values.
- **Leaky ReLU** allows a small negative output.

This small change helps gradients continue flowing during backpropagation.

---

## Program 6 - Softmax Visualization

Softmax converts raw scores into probabilities.

Example:

Raw Scores

Cat = 2.5

Dog = 1.2

Horse = 0.3

Bird = 4.8

↓

Softmax

Cat = 0.08

Dog = 0.02

Horse = 0.01

Bird = 0.89

The probabilities of all classes always sum to **1**, making Softmax ideal for **Multi-Class Classification**.

---

## Program 7 - All Activation Functions Comparison

This final visualization combines all activation functions into a single graph.

After studying this graph, you should be able to identify:

- Which activation function compresses values.
- Which one is most suitable for hidden layers.
- Which one solves the Dying ReLU problem.
- Which one is commonly used for Binary and Multi-Class Classification.

---

# Key Takeaways

- **Sigmoid** → Binary Classification
- **ReLU** → Hidden Layers
- **Leaky ReLU** → Solves Dying ReLU
- **Softmax** → Multi-Class Classification
- Understanding the graphs is just as important as memorizing the formulas.
