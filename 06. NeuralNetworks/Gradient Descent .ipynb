# Gradient Descent

Gradient Descent is an optimization algorithm used to minimize the loss function by updating the model's weights and biases.
Its objective is to move step-by-step toward the minimum loss so that the Neural Network makes more accurate predictions.

## Why Gradient Descent?

After calculating the loss, the Neural Network needs a method to reduce that error.
Gradient Descent computes the direction of the steepest decrease and updates the weights accordingly.
This process repeats until the model reaches a minimum loss.

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Learning Rate

The Learning Rate determines how large each update step should be during optimization.

- Small Learning Rate → Slow but stable learning.
- Large Learning Rate → Faster learning but may overshoot the minimum.

```python
x = np.linspace(-4, 4, 400)

loss = x**2

plt.figure(figsize=(8,5))

plt.plot(
    x,
    loss,
    linewidth=2.5,
    color="blue")

plt.scatter(
    2,
    4,
    color="red",
    s=70,
    label="Starting Point")

plt.title("Loss Function")
plt.xlabel("Weight")
plt.ylabel("Loss")
plt.grid(True)
plt.legend()

plt.savefig(
    "Assets/Gradient Descent/loss_function_graph.png",
    dpi=300,
    bbox_inches="tight")

plt.show()
```

## Gradient Descent Process

The optimizer gradually updates the weights.
As training progresses, the loss decreases until the model reaches the minimum point.

```python
epochs = np.arange(1,51)

loss = 1.3 * np.exp(-epochs/10) + 0.03

plt.figure(figsize=(8,5))

plt.plot(
    epochs,
    loss,
    marker = "o",
    markerfacecolor = "red",
    markeredgecolor = "red",
    markersize = 3,
    linewidth=1,
    color="blue")

plt.title("Gradient Descent Optimization")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.grid(True)

plt.savefig(
    "Assets/Gradient Descent/gradient_descent_graph.png",
    dpi=300,
    bbox_inches="tight")

plt.show()
```

## Effect of Different Learning Rates

The graph below compares three learning rates.
A very small learning rate learns slowly.
A very large learning rate may become unstable.
A moderate learning rate generally provides the best convergence.

```python
epochs = np.arange(1,51)

small_lr = 1.2*np.exp(-epochs/22)+0.05
good_lr = 1.2*np.exp(-epochs/10)+0.03
large_lr = 1.2*np.exp(-epochs/5)+0.12

plt.figure(figsize=(8,5))

plt.plot(
    epochs,
    small_lr,
    linewidth=2,
    color = "blue",
    label="Small LR")

plt.plot(
    epochs,
    good_lr,
    linewidth=2,
    color = "red",
    label="Optimal LR")

plt.plot(
    epochs,
    large_lr,
    linewidth=2,
    color = "#0ffc03",
    label="Large LR")

plt.title("Learning Rate Comparison")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.grid(True)
plt.legend()

plt.savefig(
    "Assets/Gradient Descent/learning_rate_comparison.png",
    dpi=300,
    bbox_inches="tight")

plt.show()
```

## Key Takeaways

- Gradient Descent minimizes the loss function.
- It updates weights in the direction of decreasing loss.
- The Learning Rate controls the size of each update.
- A suitable learning rate leads to faster and more stable convergence.
- Gradient Descent is the foundation of most Neural Network optimization algorithms.
