# Loss Functions

Loss functions measure how far a model's predictions are from the actual values. During training, the optimizer minimizes this loss to improve the model's performance.

```python
import numpy as np
import matplotlib.pyplot as plt

actual = np.array([1, 2, 3, 4, 5])
predicted = np.array([1.2, 1.8, 2.7, 4.3, 4.9])

mse = np.mean((actual - predicted) ** 2)

print(f"Mean Squared Error : {mse:.4f}")
plt.figure(figsize=(8,5))

plt.plot(actual,
         marker="o",
         linewidth=2,
         label="Actual")

plt.plot(predicted,
         marker="s",
         linewidth=2,
         label="Predicted")

plt.title("Mean Squared Error")
plt.xlabel("Sample")
plt.ylabel("Value")

plt.grid(True)

plt.legend()
plt.tight_layout()
plt.savefig("Assets/Loss Functions/mse_graph.png",
            dpi=300,
            bbox_inches="tight")

plt.show()

# Lower MSE indicates better predictions.
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Binary Cross Entropy

true_labels = np.array([1, 0, 1, 1, 0])
predicted_prob = np.array([0.95, 0.20, 0.85, 0.70, 0.10])
bce = -np.mean(
    true_labels * np.log(predicted_prob)
    +
    (1 - true_labels) * np.log(1 - predicted_prob)
)

print(f"Binary Cross Entropy : {bce:.4f}")

# Categorical Cross Entropy

true_class = np.array([0, 0, 1])
predicted_class = np.array([0.05, 0.10, 0.85])
cce = -np.sum(true_class * np.log(predicted_class))

print(f"Categorical Cross Entropy : {cce:.4f}")
plt.figure(figsize=(8,5))

plt.bar(["BCE", "CCE"],
        [bce, cce],
        color=["#4CAF50", "#2196F3"],
        edgecolor="black")

plt.title("Cross Entropy Loss Comparison")
plt.ylabel("Loss Value")
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.savefig("Assets/Loss Functions/cross_entropy_graph.png",
            dpi=300,
            bbox_inches="tight")

plt.show()

# Lower Cross Entropy indicates better classification performance.
```
