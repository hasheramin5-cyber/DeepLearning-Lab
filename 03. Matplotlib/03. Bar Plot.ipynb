```python
import numpy as np
import matplotlib.pyplot as plt

products = np.array(["Laptop", "Mouse", "Keyboard", "Monitor", "Headphones"])
sales = np.array([120, 180, 95, 75, 150])
colors = ["#4CAF50", "#2196F3", "#FFC107", "#F44336", "#9C27B0"]

plt.figure(figsize=(10, 6))

bars = plt.bar(
    products,
    sales,
    color=colors,
    edgecolor="none",
    linewidth=1.2)

plt.title(
    "Product Sales Comparison",
    fontsize=18,
    fontweight="bold")
plt.xlabel(
    "Products",
    fontsize=12,
    fontweight="bold")
plt.ylabel(
    "Units Sold",
    fontsize=12,
    fontweight="bold")

plt.grid(
    axis="y",
    linestyle="--",
    alpha=0.5)

# Display value on top of each bar
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2,
             height + 3,
             f"{height}",
             ha="center",
             fontsize=10,
             fontweight="bold")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

products = np.array(["Laptop", "Mouse", "Keyboard", "Monitor", "Headphones"])
sales = np.array([120, 180, 95, 75, 150])

colors = ["#4CAF50", "#2196F3", "#FFC107", "#F44336", "#9C27B0"]

plt.figure(figsize=(12, 6))

bars = plt.barh(products,
                sales,
                color=colors,
                edgecolor="none",
                linewidth=1.2)

plt.title("Product Sales Comparison",
          fontsize=18,
          fontweight="bold")
plt.xlabel("Units Sold",
           fontsize=12,
           fontweight="bold")
plt.ylabel("Products",
           fontsize=12,
           fontweight="bold")

plt.grid(axis="x",
         linestyle="--",
         alpha=0.5)

# Display value at the end of each ba
for bar in bars:
    width = bar.get_width()
    plt.text(width + 3,
             bar.get_y() + bar.get_height() / 2,
             f"{width}",
             va="center",
             fontsize=10,
             fontweight="bold")

plt.tight_layout()
plt.show()
```
