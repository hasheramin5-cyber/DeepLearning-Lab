```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Plot Customization
# Line Colors and Styles

x = np.array([2024, 2025, 2026, 2027, 2028])
y = np.array([45, 16, 65, 67, 88])
plt.plot(x, y, marker=".")
plt.show()
```

```python
plt.plot(x, y, marker = ".",
               markersize = 30,
               markerfacecolor = "cyan",
               markeredgecolor = "cyan",
               linestyle = "dashed",
               linewidth = 1.5,
               color = "#bd0927")
```

```python
x = np.array([2024, 2025, 2026, 2027, 2028])
y1 = np.array([45, 16, 65, 67, 88])
y2 = np.array([31, 18, 45, 23, 57])

# We can also use dictionary to save the line_style for Repetitive use.
line_style = dict(marker = ".",
                  markersize = 10,
                  markerfacecolor = "cyan",
                  markeredgecolor = "cyan",
                  linestyle = "solid",
                  linewidth = 1.5,
                  color = "#bd0927")

plt.plot(x, y1, **line_style)
plt.plot(x, y2, **line_style)

plt.show()
```

```python
x = np.array([2024, 2025, 2026, 2027, 2028])
y1 = np.array([45, 35, 65, 67, 88])
y2 = np.array([31, 46, 45, 23, 57])
y3 = np.array([51, 26, 56, 40, 80])

# We can also use dictionary to save the line_style for Repetitive use.
line_style = dict(marker = ".",
                  markersize = 5,
                  markerfacecolor = "black",
                  markeredgecolor = "gray",
                  linestyle = "solid",
                  linewidth = 1.5)

plt.plot(x, y1, color = "#FF0000", **line_style)
plt.plot(x, y2, color = "#00FF00", **line_style)
plt.plot(x, y3, color = "#0000FF", **line_style)

plt.show()
```

```python
# Final Boss
# Complete Plot Customization Example

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import MultipleLocator

x = np.array([2024, 2025, 2026, 2027, 2028])

y1 = np.array([45, 35, 65, 67, 88])
y2 = np.array([31, 46, 45, 23, 57])
y3 = np.array([51, 26, 56, 40, 80])

line_style = dict(
    marker="o",
    markersize=6,
    markerfacecolor="white",
    markeredgecolor="black",
    linestyle="--",
    linewidth=2
)

plt.figure(figsize=(10, 6))

plt.title(
    "Line Chart",
    fontsize=20,
    fontweight="bold",
    color="#00C5CC"
)

plt.xlabel(
    "Year",
    fontsize=12,
    fontweight="bold"
)

plt.ylabel(
    "Quantity",
    fontsize=12,
    fontweight="bold"
)

plt.xticks(x)
plt.yticks(np.arange(0, 101, 10))

ax = plt.gca()

# Major & Minor Grid
ax.xaxis.set_major_locator(MultipleLocator(1))
ax.xaxis.set_minor_locator(MultipleLocator(0.5))

ax.yaxis.set_major_locator(MultipleLocator(10))
ax.yaxis.set_minor_locator(MultipleLocator(5))

plt.grid(
    which="major",
    linestyle="--",
    linewidth=0.8,
    color="gray",
    alpha=0.6
)

plt.grid(
    which="minor",
    linestyle=":",
    linewidth=0.5,
    color="lightgray",
    alpha=0.8
)

plt.plot(x, y1, color="red", label="Product A", **line_style)
plt.plot(x, y2, color="green", label="Product B", **line_style)
plt.plot(x, y3, color="blue", label="Product C", **line_style)

plt.legend()

plt.tight_layout()
plt.show()
```
