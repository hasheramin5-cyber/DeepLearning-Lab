```python
# Basic Histogram

import numpy as np
import matplotlib.pyplot as plt

data = np.random.randint(1, 101, 500)
plt.hist(data)
plt.title("Basic Histogram")
plt.show()
```

```python
# Number of Bins

plt.hist(data, bins=20)
plt.title("Histogram with 20 Bins")
plt.show()
```

```python
# Colors

plt.hist(data,
         bins=20,
         color="skyblue")
plt.title("Colored Histogram")
plt.show()
```

```python
# Edge Colors
plt.hist(data,
         bins=20,
         color="skyblue",
         edgecolor="black")

plt.title("Histogram with Edge Color")
plt.show()
```

```python
# Mutiple Histogram
data1 = np.random.normal(60, 10, 500)
data2 = np.random.normal(75, 12, 500)

plt.figure(figsize=(10,6))

plt.hist(data1,
         bins=20,
         alpha=0.6,
         label="Class A")

plt.hist(data2,
         bins=20,
         alpha=0.6,
         label="Class B")

plt.legend()
plt.title("Multiple Histograms")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

scores = np.random.normal(70, 12, 1000)
plt.figure(figsize=(10,6))

# Histogram
plt.hist(scores,
         bins=20,
         color="#4CAF50",
         edgecolor="black",
         linewidth=1.2,
         alpha=0.8)

# Labels
plt.title("Student Score Distribution",
          fontsize=18,
          fontweight="bold")
plt.xlabel("Scores",
           fontsize=12,
           fontweight="bold")
plt.ylabel("Frequency",
           fontsize=12,
           fontweight="bold")

# Grid Lines
plt.grid(axis="y",
         linestyle="--",
         alpha=0.5)

plt.tight_layout()
plt.show()
```
