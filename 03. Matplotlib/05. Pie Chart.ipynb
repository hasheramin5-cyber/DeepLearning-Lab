```python
import matplotlib.pyplot as plt
import numpy as np

# Basic Pie Chart
sizes = [30, 25, 20, 15, 10]
plt.pie(sizes)
plt.title("Basic Pie Chart")
plt.show()
```

```python
# Labels

labels = ["Python", "Java", "C++", "JavaScript", "Go"]
plt.pie(sizes, labels=labels)
plt.title("Pie Chart with Labels")
plt.show()
```

```python
# Colors

colors = ["#4CAF50", "#2196F3", "#FFC107", "#F44336", "#9C27B0"]
plt.pie(sizes,
        labels=labels,
        colors=colors)

plt.title("Colored Pie Chart")
plt.show()
```

```python
# Percentages
plt.pie(sizes,
        labels=labels,
        colors=colors,
        autopct="%1.1f%%")

plt.title("Pie Chart with Percentages")
plt.show()
```

```python
# Explode

explode = [0, 0.1, 0, 0, 0]
plt.pie(sizes,
        labels=labels,
        colors=colors,
        explode=explode,
        autopct="%1.1f%%")

plt.title("Exploded Pie Chart")
plt.show()
```

```python
# Shadow

plt.pie(sizes,
        labels=labels,
        colors=colors,
        autopct="%1.1f%%",
        shadow=True)

plt.title("Pie Chart with Shadow")
plt.show()
```

```python
# Start Angle

plt.pie(sizes,
        labels=labels,
        colors=colors,
        autopct="%1.1f%%",
        startangle=90)

plt.title("Pie Chart with Start Angle")
plt.show()
```

```python
import matplotlib.pyplot as plt

labels = ["Python", "Java", "C++", "JavaScript", "Go"]
sizes = [35, 25, 18, 12, 10]
colors = ["#4CAF50", "#2196F3", "#FFC107", "#F44336", "#9C27B0"]

explode = [0.05, 0, 0, 0, 0]
plt.figure(figsize=(8,8))

plt.pie(sizes,
        labels=labels,
        colors=colors,
        explode=explode,
        autopct="%1.1f%%",
        shadow=True,
        startangle=90)

plt.title("Programming Language Popularity",
          fontsize=18,
          fontweight="bold")
plt.legend(title="Languages", loc="best")

plt.tight_layout()
plt.show()
```
