```python
# Let's Make it Complex Users !!

crop_photo = photo[145:335, 475:560]

mask_80 = np.where(crop_photo > 80, 255, 0)
mask_150 = np.where(crop_photo > 150, 255, 0)

plt.figure(figsize=(8,4))

plt.subplot(1,2,1)
plt.imshow(mask_80, cmap="gray")
plt.title("Threshold = 80")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(mask_150, cmap="gray")
plt.title("Threshold = 150")
plt.axis("off")

plt.show()
```
