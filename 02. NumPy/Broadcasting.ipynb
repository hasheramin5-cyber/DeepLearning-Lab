```python
import numpy as np

# Broadcasting
array1 = np.array([[1, 2, 3, 4]])
array2 = np.array([[1],[2],[3],[4]])
```

```python
print(array1.shape)
print(array2.shape)
```

```python
print(array1 * array2)  # The Dimensions of both have to be Same or One of the Dimensions have the size of 1.
```
