```python
import numpy as np

# Scalar Array
array1 = np.array([1, 2, 3, 4])

print(array1 + 1)
print(array1 - 2)
print(array1 * 3)
print(array1 / 4)
print(array1 ** 5)
```

```python
# Vector Math Functions
array1 = np.array([9.9867, 6.0998, 2.1125])

print(np.sqrt(array1))
print(np.round(array1))
print(np.floor(array1))
print(np.ceil(array1))
print(np.pi)
```

```python
radii = np.array([1, 2, 3])
print(np.pi*(radii ** 2))
```

```python
# Element-wise Arithmetic

array1 = np.array([1, 2, 3])
array2 = np.array([4, 5, 6])

print(array1 + array2)
print(array1 - array2)
print(array1 * array2)
print(array1 / array2)
print(array1 ** array2)
```

```python
# Comparison Operators
scores = np.array([23, 76, 89, 61, 100, 14, 69, 95])
```

```python
print(scores == 100)
```

```python
print(scores >=60 )
```

```python
scores[scores < 60] = 0
print(scores)
```

```python
# Final >_<
import numpy as np

array = np.array([1, 2, 3])

print("Original Array    :", array)
print("Squared Values    :", array ** 2)
print("Cubed Values      :", array ** 3)
print("Square Root       :", np.sqrt(array))

print()

print("Sine Values       :", np.sin(array))
print("Cosine Values     :", np.cos(array))
print("Exponential       :", np.exp(array))
print("Natural Log       :", np.log(array))
```
