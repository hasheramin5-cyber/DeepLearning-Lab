```python
import numpy as np

numbers = np.arange(1, 13)

print("Original Array:")
print(numbers)
print("Shape:", numbers.shape)
```

```python
# Reshape (1D → 2D)
matrix = numbers.reshape(3, 4)

print(matrix)
print("Shape:", matrix.shape)
```

```python
# Different Shapes

print(numbers.reshape(2, 6))
print()
print(numbers.reshape(4, 3))
print()
print(numbers.reshape(6, 2))
```

```python
# Automatic Dimension (-1)

print(numbers.reshape(3, -1))
print()
print(numbers.reshape(-1, 4))
print()
print(numbers.reshape(-1, 2))
```

```python
# Squeeze
array = np.array([[[1],[2],[3]]])

print(array.shape)
print(np.squeeze(array))
print(np.squeeze(array).shape)
```

```python
numbers = np.arange(1, 25)

print("Original Shape :", numbers.shape)

matrix = numbers.reshape(4, 6)
print()

print("Original Matrix:")
print(matrix)
print()

print("Transpose:")
print(matrix.T)
print()

print("Flatten:")
print(matrix.flatten())
print()

print("Reshape: (2,12)")
print(numbers.reshape(2,12))
print()

print("Reshape: (3,8)")
print(numbers.reshape(3,8))
```
